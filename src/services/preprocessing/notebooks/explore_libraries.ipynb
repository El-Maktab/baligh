# Libraries Exploration

This notebook verifies the installation and functionality of **Farasa** and **CAMeL Tools** for the Baligh preprocessing pipeline.

## 1. Test Farasa Segmenter


```python
from farasa.segmenter import FarasaSegmenter

# initial Farasa Segmenter
# dowload the farasa segmenter
segmenter = FarasaSegmenter(interactive=True)

text = "ذهب الطلاب إلى المدرسة"
segmented_text = segmenter.segment(text)

print(f"Original:  {text}")
print(f"Segmented: {segmented_text}")
```

## 2. Test CAMeL Tools morphology analyzer


```python
from camel_tools.morphology.analyzer import Analyzer
from camel_tools.morphology.database import MorphologyDB

# load morpholgy database
db = MorphologyDB.builtin_db()
analyzer = Analyzer(db)

word = "الطلاب"
analyses = analyzer.analyze(word)

print(f"Analyse for '{word}':")
for idx, val in enumerate(analyses[:3]):
    pos = val.get("pos")
    lemma = val.get("lex")
    diac = val.get("diac")
    print(f"  Candidate {idx + 1}: POS={pos}, Lemma={lemma}, Diac={diac}")
```

## 3. Test CAMeL Tools MLE Disambiguator


```python
from camel_tools.disambig.mle import MLEDisambiguator

# load mle disambiguator
mle = MLEDisambiguator.pretrained()

sentence = ["ذهب", "الطلاب", "إلى", "المدرسة"]
disambiguated = mle.disambiguate(sentence)

print(f"Disambiguated sentence: {sentence}")
for word in disambiguated:
    top_analysis = word.analyses[0].analysis
    pos = top_analysis.get("pos")
    lemma = top_analysis.get("lex")
    diac = top_analysis.get("diac")
    print(f"  Word: {word.word} -> POS={pos}, Lemma={lemma}, Diac={diac}")
```
