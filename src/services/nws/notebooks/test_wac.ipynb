```python
import sys
import os
from pathlib import Path

# Dynamically find the project root by searching upwards for pyproject.toml
current_dir = Path.cwd()
while current_dir.name and not (current_dir / 'pyproject.toml').exists():
    current_dir = current_dir.parent
project_root = str(current_dir)

if project_root not in sys.path:
    sys.path.append(project_root)
os.chdir(project_root)

from src.services.nws.features.wac.char_ngram.serializer import load_model
from src.services.nws.features.wac.char_ngram.model import CharNGramLM

model_path = "src/services/nws/data/char_ngram_lm.msgpack.gz"
model_data = load_model(model_path)
model = CharNGramLM(model_data)
```

```python
prefix = "انا اس"
predictions = model.predict(prefix, top_k=10)

for i, word in enumerate(predictions):
    print(f"{i+1}. {word}")
```
