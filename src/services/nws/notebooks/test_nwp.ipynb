```python
import os
import sys
from pathlib import Path

# Dynamically find the project root by searching upwards for pyproject.toml
current_dir = Path.cwd()
while current_dir.name and not (current_dir / "pyproject.toml").exists():
    current_dir = current_dir.parent
project_root = str(current_dir)

if project_root not in sys.path:
    sys.path.append(project_root)
os.chdir(project_root)

from src.services.nws.features.nwp.word_ngram.model import WordNGramLM
from src.services.nws.features.nwp.word_ngram.serializer import load_ngram_model
from src.services.nws.features.nwp.word_ngram.tokenizer import tokenize_text

model_path = "src/services/nws/data/word_ngram_lm_books.msgpack.gz"
model_data = load_ngram_model(model_path)
model = WordNGramLM(model_data)
```

```python
text = "انا عائد من العمل الى"
tokens = tokenize_text(text, is_inference=True)

predictions = model.predict_next(tokens, top_k=100)

print(f"Context: {' '.join(tokens)}")
for i, word in enumerate(predictions):
    print(f"{i + 1}. {word}")
```
