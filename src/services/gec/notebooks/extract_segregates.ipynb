```python
import sys
from pathlib import Path

project_root = Path.cwd().parents[3]
sys.path.insert(0, str(project_root))
```

```python
import json
from src.services.gec.modules.edit_tagger.common import ProjectedExample
from src.services.gec.config import NOPNX_TRAIN_OUTPUT, PNX_TRAIN_OUTPUT, CHECKPOINT_PATH
from src.services.gec.features.exporter import DatasetExporter
from src.services.gec.modules.edit_tagger.preprocessing.segregator import Segregator

def load_examples(jsonl_path):
    examples = [] 
    with open(jsonl_path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue 
            data = json.loads(line)
            examples.append(ProjectedExample(subwords=data.get("subwords", []),
                                            labels=data.get("labels", []), 
                                            labels_star=data.get("labels_star", [])))
    return examples
```

```python

segregator = Segregator()
exporter = DatasetExporter()
examples = load_examples(CHECKPOINT_PATH)

punc_examples, non_punc_examples = segregator.segregate(examples)
exporter.export_jsonl(punc_examples, NOPNX_TRAIN_OUTPUT)
exporter.export_jsonl(non_punc_examples, PNX_TRAIN_OUTPUT)
```
